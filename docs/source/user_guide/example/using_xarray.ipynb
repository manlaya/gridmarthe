# Using xarray with gridmarthe

`gridmarthe` load and store `xarray.Dataset` object.
Hence, it possible (recommended) to use `xarray` method to manipulate the data.

Other functions are provided when not found directly in `xarray` or as wrapper
methods.

Documentation: [https://docs.xarray.dev/](https://docs.xarray.dev/en/stable/index.html)
